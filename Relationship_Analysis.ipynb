```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create sample dataset
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8, 2.5, 5.5,
                    3.5, 6.5, 7.5, 4.5, 9],
    "Marks": [35, 40, 45, 52, 60, 65, 72, 85, 43, 68,
              50, 75, 80, 58, 92]
}

df = pd.DataFrame(data)

# 2. Display the dataset
print("Dataset:")
print(df)

# 3. Calculate correlation
correlation = df["Study_Hours"].corr(df["Marks"])

print("\nCorrelation Coefficient:", round(correlation, 3))

# 4. Create scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["Study_Hours"], df["Marks"], s=70)

plt.title("Relationship Analysis: Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 5. Analyze the relationship
print("\nRELATIONSHIP ANALYSIS")

if correlation > 0:
    print("The variables have a positive relationship.")
elif correlation < 0:
    print("The variables have a negative relationship.")
else:
    print("There is no linear correlation.")

print("\nOBSERVATIONS:")
print("1. Marks generally increase as study hours increase.")
print("2. The scatter plot shows an upward trend.")
print("3. The correlation coefficient indicates the direction and strength of the linear relationship.")
print("4. Correlation does not necessarily imply causation.")
```
